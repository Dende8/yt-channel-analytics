```python
from pathlib import Path

import pandas as pd

# Carpeta con los CSV generados por `transform`
CSV_DIR = Path("../data/csv")

videos = pd.read_csv(CSV_DIR / "videos.csv")
channel = pd.read_csv(CSV_DIR / "channel.csv")
videos["published_at"] = pd.to_datetime(videos["published_at"], utc=True)

# --- 2.1 Reparto por estado de privacidad ----------------------------------
# dropna=False muestra también los vídeos sin valor, si los hubiera
reparto = videos["privacy_status"].value_counts(dropna=False)

print("== 2.1 Vídeos por privacy_status ==")
print(reparto.to_string())

# --- 2.2 Públicos frente a video_count del canal ---------------------------
n_publicos = int((videos["privacy_status"] == "public").sum())
video_count_canal = int(channel.loc[0, "video_count"])

print("\n== 2.2 Públicos vs video_count del canal ==")
print(f"Vídeos públicos en videos.csv: {n_publicos}")
print(f"video_count (channel.csv):     {video_count_canal}")
print(f"Diferencia (públicos - video_count): {n_publicos - video_count_canal}")

# --- 2.3 Detalle de los vídeos no públicos ---------------------------------
no_publicos = videos[videos["privacy_status"] != "public"].sort_values(
    "published_at", ascending=False
)

print("\n== 2.3 Vídeos no públicos ==")
print(f"Cantidad: {len(no_publicos)}")
if not no_publicos.empty:
    columnas = [
        "video_id", "title", "privacy_status", "published_at",
        "format", "duration_seconds", "view_count",
    ]
    print(no_publicos[columnas].to_string(index=False))

    # Peso de los no públicos sobre el total de vistas (Data API)
    peso = no_publicos["view_count"].sum() / videos["view_count"].sum()
    print(f"\nVistas de los no públicos sobre el total: {peso:.2%}")
```
